TAHAPAN AWAL

```python
print("=" * 50)
print("IMPORT LIBRARY")
print("=" * 50)

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from scipy.stats import norm

print("Library berhasil diimpor")
```

```python
print("=" * 50)
print("MEMBACA DATASET")
print("=" * 50)

df = pd.read_csv("/content/retail_sales.csv")
display(df.head())
```

```python
print("=" * 50)
print("INFORMASI DATASET")
print("=" * 50)

print("Jumlah baris dan kolom:")
print(df.shape)

print("\nInformasi dataset:")
df.info()

print("\nNama kolom:")
print(df.columns.tolist())

print("\nData kosong:")
print(df.isnull().sum())

print("\nData duplikat:")
print(df.duplicated().sum())
```

```python
print("=" * 50)
print("PENANGANAN MISSING VALUE")
print("=" * 50)

# Kolom kategorikal menggunakan modus
df["Category"] = df["Category"].fillna(
    df["Category"].mode()[0]
)

df["Region"] = df["Region"].fillna(
    df["Region"].mode()[0]
)

# Kolom numerik menggunakan median
df["Sales"] = df["Sales"].fillna(
    df["Sales"].median()
)

df["Quantity"] = df["Quantity"].fillna(
    df["Quantity"].median()
)

print("Missing value berhasil ditangani!")
```

```python
print("=" * 50)
print("PEMERIKSAAN DATA SESUDAH PENANGANAN")
print("=" * 50)

print("Missing Value Setelah Penanganan:")
print(df.isnull().sum())

print("\nTotal Missing Value:")
print(df.isnull().sum().sum())
```

SAMPLE YANG DIAMBIL SEBANYAK 100

```python
print("=" * 50)
print("MENGAMBIL 100 SAMPLE")
print("=" * 50)

# Memastikan jumlah data mencukupi
assert len(df) >= 100, "Data kurang dari 100 sampel!"

# Mengambil 100 sampel secara acak
sampel = df.sample(
    n=100,
    random_state=42
).copy()

# Mengambil kolom Sales
data = sampel["Sales"].to_numpy()

print("Jumlah sampel:", len(data))
print("Variabel yang dianalisis: Sales")

display(sampel.head(10))
```

```python
print("=" * 50)
print("STATISTIK DESKRIPTIF DATA PENJUALAN")
print("=" * 50)

print("Jumlah sampel:", len(data))
print("Rata-rata:", np.mean(data))
print("Median:", np.median(data))
print("Nilai minimum:", np.min(data))
print("Nilai maksimum:", np.max(data))
print("Standar deviasi:", np.std(data, ddof=1))
```

BAGIAN A : MENGHITUNG ESTIMASI TITIK DAN INTERVAL MENGGUNAKAN DISTRIBUSI Z DENGAN TINGKAT KESALAHAN 5%

```python
print("=" * 50)
print("ESTIMASI TITIK")
print("=" * 50)

# Menghitung jumlah sampel
n = len(data)

# Menghitung estimasi titik
mean_data = np.mean(data)

print("ESTIMASI TITIK")

print("Jumlah sampel:", n)
print("Rata-rata penjualan:", mean_data)
print("Estimasi titik:", round(mean_data, 2))
```

```python
print("=" * 50)
print("ESTIMASI INTERVAL Z")
print("=" * 50)

# Tingkat kesalahan 5%
alpha = 0.05

# Nilai Z untuk tingkat kepercayaan 95%
z = norm.ppf(1 - alpha/2)

# Menghitung standar deviasi sampel
sigma = np.std(data, ddof=1)

# Menghitung standard error
SE = sigma / np.sqrt(n)

# Menghitung batas interval
lower = mean_data - z * SE
upper = mean_data + z * SE

print("Tingkat kepercayaan: 95%")
print("Nilai Z:", z)
print("Rata-rata:", mean_data)
print("Standar deviasi:", sigma)
print("Standard Error:", SE)

print("\nBatas bawah:", lower)
print("Batas atas:", upper)

print("\nInterval Kepercayaan 95%:")
print(f"({lower:.2f}, {upper:.2f})")
```

BAGIAN B : LAKUKAN PROSES RESAMPLING MENGGUNAKAN METODE JACKKNIFE DAN BOOTSTRAPPING, HITUNG NILAI RATA-RATA DARI HASIL RESAMPLINGNYA. UNTUK BOOTSTRAPPING LAKUKAN LITERASI SEBANYAK 1000

```python
print("=" * 50)
print("METODE JACKKNIFE")
print("=" * 50)

# Menyiapkan hasil Jackknife
jackknife_results = []

# Proses Jackknife
for i in range(n):

    # Menghilangkan satu data
    sample = np.delete(data, i)

    # Menghitung rata-rata sampel
    mean_sample = np.mean(sample)

    jackknife_results.append(mean_sample)

# Menghitung rata-rata hasil Jackknife
jackknife_mean = np.mean(jackknife_results)

# Menghitung standard error Jackknife
jackknife_se = np.sqrt(
    ((n - 1) / n) *
    np.sum(
        (jackknife_results - jackknife_mean) ** 2
    )
)
print("Jumlah resampling:", len(jackknife_results))
print("Rata-rata data awal:", mean_data)
print("Rata-rata Jackknife:", jackknife_mean)
print("Standard Error Jackknife:", jackknife_se)
```

```python
print("=" * 50)
print("METODE BOOTSTRAPPING")
print("=" * 50)

# Mengatur random seed
np.random.seed(42)

# Menyiapkan hasil Bootstrap
bootstrap_results = []

# Melakukan 1000 iterasi
for i in range(1000):

    # Mengambil sampel ulang dengan pengembalian
    sample = np.random.choice(
        data,
        size=n,
        replace=True
    )

    # Menghitung rata-rata setiap sampel
    mean_sample = np.mean(sample)

    bootstrap_results.append(mean_sample)

# Menghitung rata-rata hasil Bootstrap
bootstrap_mean = np.mean(bootstrap_results)

# Menghitung standard error Bootstrap
bootstrap_se = np.std(
    bootstrap_results,
    ddof=1
)
print("Jumlah iterasi:", len(bootstrap_results))
print("Rata-rata data awal:", mean_data)
print("Rata-rata Bootstrap:", bootstrap_mean)
print("Standard Error Bootstrap:", bootstrap_se)
```

```python
print("=" * 50)
print("PERBANDINGAN HASIL ANALISIS")
print("=" * 50)

hasil = pd.DataFrame({
    "Metode": [
        "Estimasi Titik",
        "Jackknife",
        "Bootstrapping"
    ],

    "Rata-rata": [
        mean_data,
        jackknife_mean,
        bootstrap_mean
    ],

    "Standard Error": [
        SE,
        jackknife_se,
        bootstrap_se
    ]
})

display(hasil.round(2))
```

```python
print("=" * 50)
print("GRAFIK BOOTSTARPPING")
print("=" * 50)

plt.figure(figsize=(10, 5))

plt.hist(
    bootstrap_results,
    bins=30,
    edgecolor="black"
)

plt.axvline(
    mean_data,
    linestyle="--",
    label="Rata-rata Data Awal"
)

plt.axvline(
    bootstrap_mean,
    linestyle="--",
    label="Rata-rata Bootstrap"
)

plt.title("Distribusi Hasil Bootstrapping")
plt.xlabel("Rata-rata Sales")
plt.ylabel("Frekuensi")

plt.legend()
plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python
print("=" * 50)
print("GRAFIK PERBANDINGAN")
print("=" * 50)

plt.figure(figsize=(8, 5))

plt.bar(
    hasil["Metode"],
    hasil["Rata-rata"]
)

plt.title("Perbandingan Rata-rata Setiap Metode")
plt.xlabel("Metode Analisis")
plt.ylabel("Rata-rata Sales")

plt.xticks(rotation=15)
plt.grid(axis="y", alpha=0.3)

plt.show()
```
